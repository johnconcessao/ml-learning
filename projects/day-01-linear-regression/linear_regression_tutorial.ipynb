## Step 1: Import Libraries and Load Model

````python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append('./src')
from linear_regression import LinearRegression, calculate_r2_score, plot_predictions_1d

# For reproducibility
np.random.seed(42)

print("✓ Libraries loaded successfully!")
````

## Step 2: Generate Synthetic Data

We'll create fake data where we KNOW the true relationship.
This lets us verify our model learns correctly!

**True relationship**: Price = 3 × Size + 2 (plus some random noise)

````python
# Generate 100 house samples
n_samples = 100

# X: House sizes (ranging roughly 0-4, representing scaled square footage)
X = 2.5 * np.random.randn(n_samples, 1) + 1.5

# y: House prices (in $1000s)
# True formula: y = 3*X + 2 + noise
y = 3 * X.squeeze() + 2 + np.random.randn(n_samples) * 0.5

print(f"Created {n_samples} samples")
print(f"X shape: {X.shape} (100 houses, 1 feature each)")
print(f"y shape: {y.shape} (100 prices)")
print(f"\nFirst 5 samples:")
print(f"  Size  |  Price")
print("-" * 20)
for i in range(5):
    print(f"  {X[i,0]:.2f}  |  ${y[i]:.2f}k")
````

### Visualize the Raw Data

````python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, alpha=0.6, s=50)
plt.xlabel('House Size (scaled)', fontsize=12)
plt.ylabel('House Price ($1000s)', fontsize=12)
plt.title('Raw Data: House Size vs Price', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()

print("Notice the linear trend? That's what our model will learn!")
````

## Step 3: Create and Train the Model

### What's Happening During Training?
1. **Initialize**: Start with random weights (slope=0, intercept=0)
2. **Predict**: Use current weights to make predictions
3. **Calculate Error**: How far off are we? (Mean Squared Error)
4. **Update**: Adjust weights to reduce error (Gradient Descent)
5. **Repeat**: Do this 1000 times!

This is exactly how neural networks learn, just with more layers!

````python
# Create model with:
# - learning_rate: how big a step to take (0.01 = small, safe steps)
# - n_iterations: how many times to update weights (1000 = plenty)
model = LinearRegression(learning_rate=0.01, n_iterations=1000)

print("Training model...\n")
model.fit(X, y)
````

### Understanding the Output

- **Loss decreasing**: Model is learning! ✓
- **Learned weight ≈ 3.0**: Close to true value (3.0)! ✓
- **Learned bias ≈ 2.0**: Close to true value (2.0)! ✓

If loss isn't decreasing, try:
- Lower learning rate (smaller steps)
- More iterations (more time to learn)

## Step 4: Visualize Learning Progress

Let's see how the loss changed over time

````python
model.plot_loss()
````

**What to look for:**
- **Good**: Loss starts high, drops quickly, then flattens
- **Bad**: Loss increases → learning rate too high
- **Bad**: Loss still dropping at end → need more iterations

## Step 5: Make Predictions

````python
# Predict on training data
y_pred = model.predict(X)

print("Predictions on first 10 samples:")
print(f"  Size  | Actual | Predicted | Error")
print("-" * 45)
for i in range(10):
    error = abs(y[i] - y_pred[i])
    print(f"  {X[i,0]:.2f}  | ${y[i]:.2f}k | ${y_pred[i]:.2f}k   | {error:.2f}")
````

### Predict on New Data

This is the real test - can our model generalize?

````python
# New house sizes
new_sizes = np.array([[0.5], [1.0], [2.0], [3.0], [4.0]])
new_predictions = model.predict(new_sizes)

print("Predictions for new houses:")
print(f"  Size  | Predicted Price")
print("-" * 30)
for size, price in zip(new_sizes, new_predictions):
    print(f"  {size[0]:.1f}   | ${price:.2f}k")

print("\nMakes sense? Bigger houses → higher prices!")
````

## Step 6: Evaluate Model Performance

### R² Score (R-squared)
- **1.0**: Perfect predictions
- **0.9+**: Excellent
- **0.7-0.9**: Good
- **< 0.7**: Needs improvement

````python
r2 = calculate_r2_score(y, y_pred)

print(f"Model Performance:")
print(f"  R² Score: {r2:.4f}")

if r2 > 0.9:
    print("  Rating: ⭐⭐⭐⭐⭐ Excellent!")
elif r2 > 0.8:
    print("  Rating: ⭐⭐⭐⭐ Very Good")
elif r2 > 0.7:
    print("  Rating: ⭐⭐⭐ Good")
else:
    print("  Rating: ⭐⭐ Needs Improvement")

print(f"\nInterpretation: Our model explains {r2*100:.1f}% of variance in house prices!")
````

## Step 7: Visualize Final Results

````python
plot_predictions_1d(X, y, model, "House Size (scaled)", "House Price ($1000s)")
````

**Analysis:**
- Blue dots = actual data
- Red line = our learned model
- Line should pass through the middle of the points
- Some points off the line = noise (expected!)

## Step 8: Compare to True Parameters

Remember, we created data with: **y = 3X + 2**

Did our model recover the true parameters?

````python
print("Parameter Comparison:")
print("=" * 50)
print(f"\nSlope (weight):")
print(f"  True value:    3.0000")
print(f"  Learned value: {model.weights[0]:.4f}")
print(f"  Error:         {abs(3.0 - model.weights[0]):.4f}")

print(f"\nIntercept (bias):")
print(f"  True value:    2.0000")
print(f"  Learned value: {model.bias:.4f}")
print(f"  Error:         {abs(2.0 - model.bias):.4f}")

print("\n" + "=" * 50)
if abs(3.0 - model.weights[0]) < 0.1 and abs(2.0 - model.bias) < 0.1:
    print("✓ SUCCESS! Model recovered true parameters!")
else:
    print("⚠ Close, but not perfect. This is normal due to noise!")
````

### Core ML Concepts
- **Loss Function**: Measures how wrong we are (MSE)
- **Gradient Descent**: Iteratively improves weights
- **Learning Rate**: Controls step size in optimization
- **Convergence**: Loss stops decreasing
- **Evaluation**: R² score measures fit quality

- Neural networks (just more layers!)
- Deep learning (same gradient descent)
- All modern ML (backpropagation is gradient descent)
