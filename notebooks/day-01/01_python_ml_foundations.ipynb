## 1. NumPy Basics - The Foundation of ML

NumPy (Numerical Python) is THE fundamental library for ML. Why?
- Fast operations on large arrays (written in C)
- Broadcasting for efficient computation
- Foundation for TensorFlow, PyTorch, scikit-learn

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
# This ensures we get same "random" numbers every time
np.random.seed(42)

print(f"NumPy version: {np.__version__}")
```

### Creating Arrays - Your Data Containers

```python
# 1D array - like a list of numbers (e.g., temperatures, prices)
temperatures = np.array([20, 22, 19, 25, 23])
print("1D Array (temperatures):", temperatures)
print("Shape:", temperatures.shape)  # (5,) means 5 elements
print()

# 2D array - like a spreadsheet (rows × columns)
# Each row could be a data sample, each column a feature
student_scores = np.array([
    [85, 90, 78],  # Student 1: Math, Science, English
    [92, 88, 95],  # Student 2
    [78, 85, 82]   # Student 3
])
print("2D Array (student scores):")
print(student_scores)
print("Shape:", student_scores.shape)  # (3, 3) means 3 rows, 3 columns
print()

# Creating arrays with specific values
zeros = np.zeros((3, 4))  # 3×4 array of zeros (useful for initialization)
ones = np.ones((2, 3))    # 2×3 array of ones
identity = np.eye(3)      # 3×3 identity matrix (diagonal ones)

print("Zeros (3×4):")
print(zeros)
print("\nIdentity (3×3):")
print(identity)
```

### Array Operations - The ML Magic

```python
# Element-wise operations (broadcasting)
# This is MUCH faster than Python loops!

prices = np.array([10, 20, 30, 40, 50])
print("Original prices:", prices)

# Apply 20% discount to all prices (vectorized operation)
discounted = prices * 0.8
print("After 20% discount:", discounted)

# Add $5 shipping to each
final_prices = discounted + 5
print("With $5 shipping:", final_prices)
print()

# Array arithmetic
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("Array a:", a)
print("Array b:", b)
print("a + b:", a + b)      # Element-wise addition
print("a * b:", a * b)      # Element-wise multiplication (not matrix mult!)
print("a ** 2:", a ** 2)    # Square each element
```

### Linear Algebra - The Math Behind ML

Most ML is matrix operations! Understanding these is crucial.

```python
# Dot product - fundamental to neural networks!
# Example: Calculate weighted scores

scores = np.array([85, 90, 78])  # Math, Science, English
weights = np.array([0.4, 0.4, 0.2])  # Importance of each subject

# Dot product gives weighted average
final_score = np.dot(scores, weights)
print(f"Subject scores: {scores}")
print(f"Subject weights: {weights}")
print(f"Weighted final score: {final_score:.2f}")
print()

# Matrix multiplication - how neural networks compute!
# Input features × Weights = Output

# Imagine 2 students, 3 features each
X = np.array([
    [1, 2, 3],  # Student 1
    [4, 5, 6]   # Student 2
])  # Shape: (2, 3)

# Weight matrix (3 features → 2 outputs)
W = np.array([
    [0.1, 0.4],
    [0.2, 0.5],
    [0.3, 0.6]
])  # Shape: (3, 2)

# Matrix multiplication
output = np.matmul(X, W)  # or X @ W
print("Input X (2 students, 3 features):")
print(X)
print("\nWeights W (3 features → 2 outputs):")
print(W)
print("\nOutput (2 students, 2 predictions):")
print(output)
print("\nThis is exactly what happens in a neural network layer!")
```

### Statistics - Understanding Your Data

```python
# Generate some random data (like real-world measurements)
data = np.random.randn(100) * 10 + 50  # 100 values, mean≈50, std≈10

print("Data statistics:")
print(f"Mean (average): {np.mean(data):.2f}")
print(f"Median (middle value): {np.median(data):.2f}")
print(f"Std deviation (spread): {np.std(data):.2f}")
print(f"Min: {np.min(data):.2f}")
print(f"Max: {np.max(data):.2f}")
print()

# Visualize the distribution
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(data, bins=20, edgecolor='black', alpha=0.7)
plt.axvline(np.mean(data), color='red', linestyle='--', label=f'Mean: {np.mean(data):.2f}')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution of Data')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(data, marker='o', markersize=3, linestyle='-', linewidth=0.5)
plt.axhline(np.mean(data), color='red', linestyle='--', label='Mean')
plt.xlabel('Sample Index')
plt.ylabel('Value')
plt.title('Data Points Over Time')
plt.legend()

plt.tight_layout()
plt.show()

print("Understanding these statistics helps you know if your ML model is working!")
```

### Reshaping and Indexing - Data Wrangling

```python
# Create a range of numbers
arr = np.arange(12)  # [0, 1, 2, ..., 11]
print("Original array:", arr)
print("Shape:", arr.shape)
print()

# Reshape to 2D (useful for preparing data for ML)
matrix = arr.reshape(3, 4)  # 3 rows, 4 columns
print("Reshaped to 3×4 matrix:")
print(matrix)
print()

# Indexing (accessing elements)
print("First row:", matrix[0])           # Row 0
print("First column:", matrix[:, 0])     # All rows, column 0
print("Element at (1,2):", matrix[1, 2]) # Row 1, column 2 = value 6
print()

# Slicing (like Python lists, but multi-dimensional)
print("First 2 rows, first 3 columns:")
print(matrix[:2, :3])
print()

# Boolean indexing (filtering data)
print("Values greater than 5:")
print(matrix[matrix > 5])

print("\nThese operations are crucial for preparing ML datasets!")
```

### Vectorization - Why NumPy is FAST

```python
import time

# Let's compare Python loops vs NumPy vectorization
size = 1000000
a = np.random.rand(size)
b = np.random.rand(size)

# Method 1: Python loop (SLOW)
start = time.time()
result_loop = []
for i in range(size):
    result_loop.append(a[i] * b[i])
time_loop = time.time() - start

# Method 2: NumPy vectorization (FAST)
start = time.time()
result_vec = a * b
time_vec = time.time() - start

print(f"Python loop time: {time_loop:.4f} seconds")
print(f"NumPy vectorization time: {time_vec:.4f} seconds")
print(f"\nNumPy is {time_loop/time_vec:.1f}x FASTER!")
print("\nThis is why we use NumPy for ML - imagine training on millions of samples!")
```
